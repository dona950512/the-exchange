```python
n_student, n_course = 6, 5
scores = np.random.randint(50, 100, n_student * n_course).reshape(n_student, n_course)
students = ['Student {:d}'.format(i + 1) for i in range(n_student)]
courses = ['Course {:d}'.format(i + 1) for i in range(n_course)]
df = pd.DataFrame(scores, index=students, columns=courses)

print('Original')
print(df)
print()
print('Sorted by Course 2. Descending:')
df = df.sort_values(by='Course 2', ascending=False)  # 依照Couse 2排序，從大到小
print(df)
print()
print('Sorted by Course 2. Ascending:')
df = df.sort_values(by='Course 2', ascending=True)  # 依照Couse 2排序，從小到大
print(df)
```

```python
df = pd.DataFrame(
    {
        'userId': ['Jack', 'Tim', 'Jack', 'Tim', 'Jack', 'Tim', 'Jack', 'Tim'],
        'movieId': ['The Godfather', 'The Godfather', 'Braveheart', 'Braveheart', 'Apollo 13', 'Apollo 13', 'Batman', 'Batman'],
        'rating': np.random.randint(0, 6, 8)
    }
)
print(df)
print()
print('把相同用戶的評分紀錄蒐集成一個類似Series的物件，再做平均')
print(df.groupby('userId')['rating'].mean())
print()
print('把相同電影的評分紀錄蒐集成一個類似Series的物件，再做平均')
print(df.groupby('movieId')['rating'].mean())
print()
print('把相同用戶的評分紀錄蒐集成一個類似Series的物件，再取中位數')
print(df.groupby('userId')['rating'].apply(lambda series: series.median()))
```

```python
print(df)
print()
print('將userId當成索引，movieId當成欄位，rating當成內部資料，做成矩陣')
print(pd.pivot_table(df, index=["userId"], columns=["movieId"], values="rating"))
```
